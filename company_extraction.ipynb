```python
import os
import pandas as pd

# Step 1: Define paths
INPUT_FOLDER = r"D:\\Data - Driven - Stock - Analysis\\YAML_to_CSV"  # Folder with daily CSVs
OUTPUT_FOLDER = r"D:\\Data - Driven - Stock - Analysis\\segregated_csv_by_company"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Step 2: Dictionary to collect data per company
symbol_data = {}

# Step 3: Loop through all CSV files in the input folder
for filename in os.listdir(INPUT_FOLDER):
    if filename.endswith(".csv"):
        filepath = os.path.join(INPUT_FOLDER, filename)
        df = pd.read_csv(filepath)

        # Make sure column names are correct
        if 'Ticker' not in df.columns:
            print(f"❌ Missing 'Ticker' column in {filename}, skipping")
            continue

        # Loop over all rows and collect rows company-wise
        for _, row in df.iterrows():
            symbol = row['Ticker']
            symbol_data.setdefault(symbol, []).append(row)

# Step 4: Save each company’s data to a separate CSV
for symbol, rows in symbol_data.items():
    df = pd.DataFrame(rows)
    df.sort_values(by='Date', inplace=True)
    df.to_csv(os.path.join(OUTPUT_FOLDER, f"{symbol}.csv"), index=False)

print(f"✅ Done! All 50 company CSV files created in:\n{OUTPUT_FOLDER}")
```

```python
import os
import pandas as pd

# Folder containing the 50 company-wise CSVs
INPUT_FOLDER = r"D:\\Data - Driven - Stock - Analysis\\segregated_csv_by_company"

# Loop through all CSV files and drop 'Date' column
for filename in os.listdir(INPUT_FOLDER):
    if filename.endswith(".csv"):
        filepath = os.path.join(INPUT_FOLDER, filename)
        
        # Load CSV
        df = pd.read_csv(filepath)

        # Drop 'Date' column if it exists
        if 'Date' in df.columns:
            df.drop(columns=['Date'], inplace=True)
            df.to_csv(filepath, index=False)
            print(f"✅ 'Date' column removed from {filename}")
        else:
            print(f"ℹ️ No 'Date' column in {filename}, skipped.")

print("🎉 Done! All files processed.")
```
